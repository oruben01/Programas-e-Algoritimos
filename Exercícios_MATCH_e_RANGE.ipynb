```python
# 1 Exercício MATCH

cor = "azul"

match cor:
    case "azul":
        print("Cor azul")
    case _:
        print("Outra cor")
```

```python
# 2 Exercício MATCH

numero = 3

match numero:
    case 1:
        print("Um")
    case 2:
        print("Dois")
    case 3:
        print("Três")
    case _:
        print("Número inválido")
```

```python
# 3 Exercício MATCH

opcao = 1

match opcao:
    case 1:
        print("Água")
    case 2:
        print("Suco")
    case 3:
        print("Refri")
    case _:
        print("Opção inválida")
```

```python
# 4 Exercício MATCH

dia = "domingo"

match dia:
    case "sábado" | "domingo":
        print("Fim de semana")
    case _:
        print("Dia de semana")
```

```python
# 5 Exercício MATCH

nota = "C"

match nota:
    case "A":
        print("Excelente")
    case "B":
        print("Bom")
    case "C":
        print("Precisa melhorar")
    case _:
        print("Nota inválida")
```

```python
# 1 Exercício RANGE

for numero in range(10):
    print(numero)
```

```python
# 2 Exercício RANGE

for numero in range(1, 11):
    print(numero)
```

```python
# 3 Exercício RANGE

for numero in range(5, 16):
    print(numero)
```

```python
# 4 Exercício Range

for numero in range(0, 21, 2):
    print(numero)
```

```python
# 5 Exercício Range


for numero in range(1, 16, 2):
    print(numero)
```

```python
# 6 Exercício Range

for numero in range(10, 0, -1):
    print(numero)
```

```python
# 7 Exercício Range

for numero in range(1, 11):
    print(f"3 x {numero} = {3 * numero}")
```
